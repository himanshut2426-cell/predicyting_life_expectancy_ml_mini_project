```python
  df['Life expectancy '] = df['Life expectancy '].fillna(df['Life expectancy '].median(), inplace=True)
  df['Adult Mortality'] = df['Adult Mortality'].fillna(df['Adult Mortality'].median(), inplace=True)
  df['Hepatitis B'] = df['Hepatitis B'].fillna(df['Hepatitis B'].median(), inplace=True)
  df['Measles '] = df['Measles '].fillna(df['Measles '].median(), inplace=True)
  df[' BMI '] = df[' BMI '].fillna(df[' BMI '].median(), inplace=True)
  df['under-five deaths '] = df['under-five deaths '].fillna(df['under-five deaths '].median(), inplace=True)
  df[' HIV/AIDS'] = df[' HIV/AIDS'].fillna(df[' HIV/AIDS'].median(), inplace=True)
  df['Total expenditure'] = df['Total expenditure'].fillna(df['Total expenditure'].median(), inplace=True)
  df['GDP'] = df['GDP'].fillna(df['GDP'].median(), inplace=True)
  df['Population'] = df['Population'].fillna(df['Population'].median(), inplace=True)
  df[' thinness  1-19 years'] = df[' thinness  1-19 years'].fillna(df[' thinness  1-19 years'].median(), inplace=True)
  df[' thinness 5-9 years'] = df[' thinness 5-9 years'].fillna(df[' thinness 5-9 years'].median(), inplace=True)
  df['Income composition of resources'] = df['Income composition of resources'].fillna(df['Income composition of resources'].median(), inplace=True)
  df['Schooling'] = df['Schooling'].fillna(df['Schooling'].median(), inplace=True)
  df['Alcohol'] = df['Alcohol'].fillna(df['Alcohol'].median(), inplace=True)
  df['Polio'] = df['Polio'].fillna(df['Polio'].median(), inplace=True)
  df['Diphtheria '] = df['Diphtheria '].fillna(df['Diphtheria '].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.strip()
```

```python
x = df.drop(['Life expectancy','Status'],axis=1)
y = df['Life expectancy']
```

```python
import sklearn
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score ,mean_absolute_error ,mean_absolute_percentage_error 
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
model.coef_
```

```python
train_pred = model.predict(x_train)
test_pred = model.predict(x_test)
```

```python
print("Mean Squared Error (MSE):", mean_squared_error(y_test, test_pred))
print("end=Mean Absolute Error (MAE):", mean_absolute_error(y_test, test_pred))
print("mean absolute percentage error:", mean_absolute_percentage_error(y_test, test_pred))
```

```python
print("Mean Squared Error (MSE):", mean_squared_error(y_train, train_pred))
print("end=Mean Absolute Error (MAE):", mean_absolute_error(y_train, train_pred))
print("mean absolute percentage error:", mean_absolute_percentage_error(y_train, train_pred)*100)
```
